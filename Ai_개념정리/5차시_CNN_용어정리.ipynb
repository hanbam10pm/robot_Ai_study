### 1.컴퓨터 비전

시각적 데이터(이미지나 동영상)를 처리 및 분석하고 이해할 수 있는 디지털 시스템을 만드는 데 중점을 둔 컴퓨터 과학 분야

컴퓨터가 이미지를 픽셀 단위로 처리하고 이해하도록 가르치는 것에 기반

### 2.이미지 데이터의 특성

- Matrx

숫자로 이루어진 2차원 배열

행(row)과 열(column)로 이미지의 크기가 결정되는 직사각형 형태

Matrix 크기가 클수록, 필요 저장공간과 처리하는데 소요되는 리소스가 비례하여 증가

- Pixels

Matrix를 이루고 있는 숫자 데이터

각 숫자값들은 밝기를 나타냄

Matrix 크기가 클수록, 공간 해상도(spartial resultion)가 좋음

- Bit depth

픽셀이 갖는 밝기 값의 범위

범위가 클수록 이미지가 표현하는 밝기가 다양해지며, 필요로 하는 저장공간은 커짐

- Channel

각 픽셀이 갖고 있는 숫자로 이루어진 벡터

RGB(red, green, blue)에서 red channel의 의미? : 이미지 데이터에서 red밝기에 대한 값들만 모아놓은 것

### 3.CNN의 특징

- 가중치 공유

필터의 가중치를 전체 이미지에 동일하게 적용하여 학습해야 할 매개변수를 효율적으로 줄임


- 자동 특징 추출

특징 추출 단계를 여러 개 배치하여 CNN이 데이터로부터 특징 표현을 스스로 학습하게 함


- 계층적 학습

다층 구조를 통해 저수준(점,선)부터 고수준(형태, 객체)특징까지 단계적으로 학습하여 복잡한 패턴 인식 가능


- 시공간 상관관계 탐색

이미지 내 인접 픽셀 간의 공간적 관계나 비디오 데이터의 시간적 상관관계를 효과적으로 탐색하고 활용 가능

### 4.Convolution(합성곱)

데이터 위에 작은 필터를 슬라이딩하면서 원소별 곱셈 후 합산하는 연산

이미지의 엣지, 질감, 패턴같은 지역적 특징을 자동으로 추출한다.

- 필터

학습 가능한 가중치 행렬로, 3x3 크기의 작은 윈도우 같은 것.

- 스트라이드

필터가 한 번에 이동하는 칸 수. 1이면 한 칸씩, 2이면 두 칸씩 이동한다.

- 패딩

입력 가장자리에 0을 채워 출력 크기를 조절한다.

- 채널

컬러 이미지는 RGB 3채널이며, 필터도 입력 채널 수와 동일한 깊이를 가진다. 필터 개수가 곧 출력 채널 수가 된다.

- 출력 크기 공식

출력 크기 = (입력 크기 - 필터 크기 + 2 x 패딩) / 스트라이드+1

### 5.Pooling(풀링)

- max-pooling

필터가 덮고 있는 Feature Map 영역에서 최대 요소를 선택한다.

따라서 max-pooling층 이후의 출력은 이전 특징맵의 가장 두드러진 특징을 포함한다.


- Average-pooling

필터가 덮고 있는 Feature Map 영역에서 요소들의 평균을 계산한다.

따라서 max-pooling층 이후의 출력은 이전 특징맵의 요소에서 평균적인 특징을 포함한다.


### 6.CNN의 기본 아키텍쳐

* Convolution Layer -> Polling Layer -> Fully Connected Layer(FC)

- Convolution Layer

필터가 이미지 위를 지나가며 한 번에 몇 픽셀씩 스캔하고 각 특징이 속하는 클래스를 예측하는 특징맵을 만든다.

- Pooling Layer

합성곱 계층에서 얻은 각 특징의 정보를 줄이면서 가장 중요한 정보를 유지한다.


- Fully Connected Layer(FC)

완전 연결 계층 : 이전 계층의 출력을 받아 '평탄화'한 후 단일 벡터로 변환한다.



### 7.Softmax(소프트맥스)

CNN의 마지막 출력층에서 사용되며, FC계층이 출력한 임의의 실수값(logits)을 확률 분포로 변환하는 함수

어떤 실수값(logits)이든 0~1사이의 확률로 바뀌고, 모든 클래스의 확률 합은 항상 1이 된다.

결론적으로 모델의 원시 출력을 사람이 이해할 수 있는 확률로 변환해주는 역할을 한다.

### 8.AlexNet

2012년 Alex Krizhevsky 팀이 발표한 CNN모델

- ReLU를 대규모 CNN에 최초로 성공적으로 적용하여 표준 활성화 함수가 됨.

- 과적합 위험이 있어 Dropout을 적용함

- GPU를 딥러닝 학습에 활용하는 패러다임을 열었다.

### 9.GoogleNet

구글에서 발표한 Inception모듈이라는 구조를 가진 CNN모델

- 1x1conv를 통해 채널을 축소하여 연산량을 줄이고 ReLU로 비선형을 추가한다.

- 채널 축소를 통해 연산 효율을 극대화하는 bottleneck개념을 확립했다.

### 10.VGG-19-BN

옥스포드 대학의 VGGNet 연구팀이 제안한 아키텍처에 BN(Batch Normalization)을 추가한 CNN모델

- 3x3 작은 필터를 깊게 쌓자는 설계 철학을 가지고 있다.

- 19는 가중치가 있는 층(Conv+FC)가 19개라는 뜻

- Batch Norm의 역할 : 각 미니배치 단위로 정규화를 실행하여 각 층의 입력 분포를 안정적으로 만들고 학습속도가 빨라진다. 약간의 노이즈가 추가되어 과적합 방지 효과도 나타난다.


### 11.ResNet-18

마이크로소프트에서 발표한 잔차 연결을 도입하여 깊은 신경망의 학습 문제를 해결한 CNN모델

- VGG에서 단순이 층을 깊게 쌓으면 발생하는 기울기 소실문제를 잔차 학습을 통해 기울기를 앞쪽으로 전달하는 방식으로 해결

- 깊이가 깊어질수록 VGG에 비해 강력한 성능을 나타낸다.

### 12.LeNet-5

1988년 Yann LeCun등이 발표한 CNN모델, 손글씨 숫자(0~9) 인식을 위해 설계되었다.

Convolution -> Pooling -> Fully Connected라는 패턴을 최초로 확립한 모델.

- 2개의 conv층과 2개의 subsampling층, 3개의 FC층으로 구성되어 있다.

- 강건성 : 이미지의 회전, 위치 변화, 크기 변화에 대해 불변성을 가지며 왜곡에도 강한 특성을 보인다.

- 약 6만개의 파라미터로 현대 모델 대비 매우 가볍고 효율적인 구조를 가진다.

- 서브 샘플링 : 공간 크기를 줄여 연산량을 감소시킨다.(가중치를 가진 평균 풀링을 사용한다.)


- local connectivity

이미지와 같은 고차원 입력에서 현재 레이어의 한 노드를 이전 볼륨의 모든 노드들과 연결하는 것은 비효율적이다.

대신, 레이어의 각 노드들을 입력 볼륨의 로컬한 영역에만 연결한다.

이 영역을 receptive field라고 한다.

- Shared wights

convolution을 위해 filter를 적용할 때, filter가 적용된 결과는 계속 변경되지만. 적용하는 filter값은 변하지 않는 것을 의미

한마디로 동일한 weight가 convolution할 때, 동일하게 적용되는 것을 의미

- Spatial or temporal sub-sampling (Pooling)
